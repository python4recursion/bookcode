```python
# Section 2.6: shuffle1.py
# shuffle only once
def interleave(combined, leftgroup, rightgroup, results):
    # combined: the deck of shuffled cards released by two hands
    # leftgroup: the group of remaining cards held by left hand
    # rightgroup: the group of remaining cards held by right hand
    # stop conditions: either group is empty
    # if left group is empty, add all cards of the right deck
    # if right group is empty, add all cards of the left deck
    # make a copy, do not modify combined directly
    newgroup = list(combined)
    if (leftgroup == []):
        for i in range(0, len(rightgroup)):
            newgroup.append(rightgroup[i])
        results.append(newgroup)
        return
    # if right group is empty, add all cards of the left deck
    if (rightgroup == []):
        for i in range(0, len(leftgroup)):
            newgroup.append(leftgroup[i])
        results.append(newgroup)
        return
    # neither group is empty, pick one card from either left or right
    newgroup = list(combined) # make a copy
    newgroup.append(leftgroup[0]) # take one card from left
    llen = len(leftgroup)
    # exclude the first card by setting 1 as the beginning index
    interleave(newgroup, leftgroup[1:llen + 1], rightgroup, results)
    newgroup = list(combined)
    newgroup.append(rightgroup[0]) # take one card from right
    rlen = len(rightgroup)
    # the right deck has one fewer card
    interleave(newgroup, leftgroup, rightgroup[1:rlen + 1], results)
def shuffle(k):
    # create the original deck of cards
    array = ['A'] # create an array of A, 2, 3, 4, ...
    for i in range (2, k + 1):
        array.append(str(i))
    # divide the deck into left deck and right deck
    results = list()
    for i in range (1, k):
        leftgroup = array[0:i]
        rightgroup = array[i: k + 1]
        combined = []
        interleave(combined, leftgroup, rightgroup, results)
    return results
```

```python
import random
def testshuffle():
    n = random.randint(3, 9)
    # since the number of results grow exponentially,
    # consider only 3 to 9 cards
    results = shuffle(n)
    for r in results:
        print (r)
    print(f'n = {n}, number of results = {len(results)}')
testshuffle()
```
