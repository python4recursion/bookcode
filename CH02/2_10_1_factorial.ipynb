```python
# Section 2.10.1: factorial1.py
import sys
def f1(n):
  if (n == 1):
    return 1
  return n * f1(n - 1)
```

```python
def f2(n):
  if (n == 1):
    return 1
  product = 1
  for x in range(2, n + 1):
    product *= x
  return product
```

```python
import timeit
for n in range(41, 51):
  if (f1(n) != f2(n)):
    print(f'f1({n}) = {f1(n)}, f2({n}) = {f2(n)}')
  t1 = timeit.timeit(lambda: f1(n), number = 10000)
  t2 = timeit.timeit(lambda: f2(n), number = 10000)
  print(f'n = {n}, t1= {t1:.6e}, t2 = {t2:.6e}, ratio = {t1/t2:.4}')
```
