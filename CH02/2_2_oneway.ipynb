```python
# Section 2.2: oneway.py
import sys
import random
def oneway(x, y):
    if ((x < 0) or (y < 0)):
        return 0
    if ((x == 0) or (y == 0)):
        return 1
    a = oneway(x - 1, y)
    b = oneway(x, y - 1)
    return (a + b)
if __name__ == "__main__":
    n = random.randint(3, 11)
    m = random.randint(3, 11)
    options = oneway(n, m)
    print(f'(n,m) = ({n},{m}), options = {options}')
```
