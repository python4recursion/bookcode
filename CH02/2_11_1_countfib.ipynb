```python
# Section 2.11.1: countfib.py
# count the number of times the fib function is called
def fib(n, counts):
    counts[n] += 1
    if (n < 1): # invalid
        return 0
    if (n == 1):
        return 1
    if (n == 2):
        return 1
    return (fib(n-1, counts) + fib(n-2, counts))
if __name__ == "__main__":
    size = 9
    for n in range(2, size + 1):
        counts = [0 for ind in range(n + 1)]
        fib(n, counts)
        print(f'fib({n}), counts={counts}')
```
