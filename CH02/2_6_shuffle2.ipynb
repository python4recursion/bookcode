```python
# Section 2.6: shuffle2.py
# shuffle multiple times
def interleave(combined, leftgroup, rightgroup, r, results):
    newgroup = list(combined)
    if (leftgroup == []):
        for i in range(0, len(rightgroup)):
            newgroup.append(rightgroup[i])
        divide(newgroup, r - 1, results)
        return
    if (rightgroup == []):
        for i in range(0, len(leftgroup)):
            newgroup.append(leftgroup[i])
        divide(newgroup, r - 1, results)
        return
    newgroup = list(combined)
    newgroup.append(leftgroup[0])
    llen = len(leftgroup)
    interleave(newgroup, leftgroup[1:llen + 1], rightgroup, r, results)
    newgroup = list(combined)
    newgroup.append(rightgroup[0])
    rlen = len(rightgroup)
    interleave(newgroup, leftgroup, rightgroup[1:rlen + 1], r, results)
def divide(array, r, results):
    if (r == 0): # round is zero, no more shuffling
        results.append (array)
        return
    k = len(array)
    for i in range (1, k):
        leftgroup = array[0:i]
        rightgroup = array[i: k + 1]
        combined = []
        interleave(combined, leftgroup, rightgroup, r, results)
def shuffle(k, r):
    array = ['A']
    results = list()
    for i in range (2, k + 1):
        array.append(str(i))
    divide(array, r, results)
    return results
```

```python
import random
def testshuffle2():
    n = random.randint(3, 7) # number of cards
    r = random.randint(1, 5) # number of rounds
    results = shuffle(n, r)
    for res in results:
        print (res)
    print(f'n = {n}, r = {r}: number of results = {len(results)}')
    print(f'unique results = {len(set(tuple(res) for res in results))}')
testshuffle2()
```
