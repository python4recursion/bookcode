```python
# Section 2.11.9: Hanoi (4 poles)
class Counter:
  def __init__(self):
    self.count = 0
  def increment(self, num = 1):
    self.count += num
  def get(self):
    return self.count
```

```python
def move4(disk, src, dest, additional1, additional2, cnt):
  # 4 poles
  if (disk <= 0):
    print('Invalid argument')
    return
  if (disk == 1):
    cnt.increment()
    print('move disk 1 from ' + src + ' to ' + dest)
    return
  if (disk == 2):
    cnt.increment(3)
    print('move disk 1 from ' + src + ' to ' + additional1)
    print('move disk 2 from ' + src + ' to ' + dest)
    print('move disk 1 from ' + additional1 + ' to ' + dest)
    return
  move4(disk - 2, src, additional1, additional2, dest, cnt)
  cnt.increment(3)
  print('move disk ' + str(disk - 1) +  ' from ' + src + ' to ' + additional2)
  print('move disk ' + str(disk) +  ' from ' + src + ' to ' + dest)
  print('move disk ' + str(disk - 1) +  ' from ' + additional2 + ' to ' + dest)
  move4(disk - 2, additional1, dest, src, additional2, cnt)
```

```python
def testhanoi4():
    counters = {}
    for num in range(2, 7):
      cnt = Counter()
      print(f'\nnum = {num}')
      move4(num, 'A', 'B', 'C', 'D', cnt)
      counters[num] = cnt.get()
    print('\ncounters:')
    for num in counters:
      print(f'num = {num}, counter = {counters[num]}')
testhanoi4()
```
