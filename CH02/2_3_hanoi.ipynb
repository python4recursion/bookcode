```python
# Section 2.3: hanoi.py (3 poles)
class Counter:
  def __init__(self):
    self.count = 0
  def increment(self, num = 1):
    self.count += num
  def get(self):
    return self.count
def move3(disk, src, dest, additional, cnt):
  # 3 poles
  if (disk <= 0):
    print('Invalid argument')
    return
  if (disk == 1):
    cnt.increment()
    print('move disk 1 from ' + src + ' to ' + dest)
    return
  move3(disk - 1, src, additional, dest, cnt)
  cnt.increment()
  print('move disk ' + str(disk) +  ' from ' + src + ' to ' + dest)
  move3(disk - 1, additional, dest, src, cnt)
```

```python
def testhanoi3():
    counters = {}
    for num in range(2, 7):
      cnt = Counter()
      print(f'\nnum = {num}')
      move3(num, 'A', 'B', 'C', cnt)
      counters[num] = cnt.get()
    for num in counters:
      print(f'num = {num}, counter = {counters[num]}')
testhanoi3()
```
