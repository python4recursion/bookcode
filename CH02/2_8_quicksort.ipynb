```python
# Section 2.8: quicksort.py
import random
def helper(arr, first, last):
    if (first >= last):  # zero or one element, already sorted
        return
    pivot = arr[first]
    low = first + 1
    high = last
    while (low < high):
        while ((low < last) and (arr[low] <= pivot)): # notice <=
            low = low + 1
        while ((first < high) and (arr[high] > pivot)): # notice >
            high = high - 1
        if (low < high):
            arr[low], arr[high] = arr[high], arr[low] # swap
    if (pivot > arr[high]):
        arr[first], arr[high] = arr[high], arr[first] # swap
    helper(arr, first, high - 1) # sort the first part
    helper(arr, low, last) # sort the third part
def quickSort(arr):
    helper(arr,0, len(arr) - 1)
    return arr
def checkOrder(arr):
    for ind in range(0, len(arr) - 1):
        if (arr[ind] > arr[ind + 1]):
            print('Error')
if __name__ == "__main__":
    length = 100
    arr = []
    for ind in range(0, length):
        arr.append(random.randint(1, 3 * length))
    quickSort(arr)
    checkOrder(arr)
```
