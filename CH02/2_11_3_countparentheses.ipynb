```python
# Section 2.11.3: countparentheses.py
def g(n, m):
  if (n > m):
    return 0
  if ((n == 0) and (m != 0)):
    return 1
  if ((n == 0) and (m == 0)):
    return 0
  return g(n - 1, m) + g(n, m - 1)
```

```python
import random
def countparentheses():
  n = random.randint(1,8) + 1
  # print(f'g({n},{m}) = {g(n,m)}')
  print(' ',end=' | ')
  for r in range(1, n):
    print(f'{r:5}', end=' ')
  print('')
  print('-' * 7 * n)
  for m in range(1, n):
    print(m, end=' | ')
    for r in range(1, n):
      print(f'{g(r,m):5}', end=' ')
    print('')
countparentheses()
```
