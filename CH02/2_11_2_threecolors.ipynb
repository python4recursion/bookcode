```python
# Section 2.11.2: threecolorballs.py
# print the number of options for selecting balls of three colors
import sys
def r(n): # number of options to select n balls, first is red
  if (n <= 1):
    return 1
  return (b(n-1) + g(n-1))
def g(n): # number of options to select n balls, first is green
  if (n <= 1):
    return 1
  return (r(n-1) + b(n-1))
def b(n): # number of options to select n balls, first is blue
  if (n <= 1):
    return 1
  return (r(n-1) + g(n-1) + b(n-1))
def f(n): # number of options to select n balls
  return (r(n) + g(n) + b(n))
def print_options(label, func, num):
  print(f'{label}(n)', end=' | ')
  for n in range(1, num + 1):
    print(f'{func(n):5}', end=' ')
  print('')
def three_colors_balls(num):
  print('    ', end=' | ')
  for n in range(1, num + 1):
    print(f'{n:5}', end=' ')
  print('')
  print('-' * 50)
  print_options('r', r, num)
  print_options('g', g, num)
  print_options('b', b, num)
  print_options('f', f, num)
three_colors_balls(6)
```
