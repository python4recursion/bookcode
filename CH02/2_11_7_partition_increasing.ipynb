```python
# Section 2.11.7: partitionincreasing.py
# partition a number using increasing numbers
from datetime import datetime
class Counter:
    def __init__(self):
        self.value = 0
    def increment(self):
        self.value += 1
def partition1(num, usedint, ind, minval, count, results):
    # generate only valid partitions
    count.increment()
    if (num == 0): # stop condition
        results.append(usedint[0:ind])
        return
    for val in range(minval, num + 1):
        valid = False
        if (ind == 0):
            valid = True
        elif (usedint[ind - 1] < val):
            valid = True
        if (valid == True):
            usedint[ind] = val
            partition1(num - val, usedint, ind + 1, val + 1, count, results)
def partition2(num, usedint, ind, minval, count, results):
    # generate all partitions, print only valid paritions
    count.increment()
    if (num == 0):
        valid = True
        for i in range(1, ind):
            if (usedint[i - 1] >= usedint[i]):
                valid = False
        if (valid == True):
            results.append(usedint[0:ind])
        return
    for val in range(1, num + 1):
        usedint[ind] = val
        partition2(num - val, usedint, ind + 1, minval, count, results)
def measuretime(num, func, message):
    usedint = [0] * num # initialize to 0
    count = Counter()
    results = []
    start_time = datetime.now()
    func(num, usedint, 0, 1, count, results)
    end_time = datetime.now()
    print(message)
    if (num < 10):
        print(results)
    print(f'count: {count.value}')
    print(f'time: {end_time - start_time}')
def partition(num):
    measuretime(num, partition1, '--- partition 1 ---')
    measuretime(num, partition2, '--- partition 2 ---')
if __name__ == "__main__":
    n = 25
    partition(n)
```
