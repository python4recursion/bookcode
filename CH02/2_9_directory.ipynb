```python
# Section 2.9: directory.py
# list files and directories in a tree-like format
import os
def tree(dir_path, fhd, prefix = "", file_count=0, dir_count=0):
    entries = sorted(os.listdir(dir_path), key=str.casefold)
    entries_count = len(entries)
    for index, entry in enumerate(entries):
        full_path = os.path.join(dir_path, entry)
        if (index == entries_count - 1):
            connector = "`-- "
        else:
            connector = "|-- "
        if os.path.isdir(full_path):
            fhd.write(prefix + connector + entry + '\n')
            dir_count += 1
            if (index == entries_count - 1):
                extension = "    "
            else:
                extension = "|   "
            file_count, dir_count = tree(full_path, fhd, prefix + extension, file_count, dir_count)
        else:
            fhd.write(prefix + connector + entry + '\n')
            file_count += 1
    return file_count, dir_count
```

```python
!git clone https://github.com/python4recursion/bookcode.git
base_path = 'bookcode/'
myoutput_file = os.path.join(base_path, 'mytree_output.txt')
fhd = open(myoutput_file, 'w')
file_count, dir_count = tree(base_path, fhd)
fhd.write(f'\n{dir_count} directories, {file_count} files\n')
fhd.close()
```
