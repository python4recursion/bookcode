```python
# Section 2.4: partition.py
import sys
import random
def helper(num, usedint, ind, results):
    # num: the value to partition
    # usedint: array storing the integers already used
    # ind: index for the array, start from 0
    if (num == 0): # stop condition
        results.append(usedint[0:ind])
        return
    for val in range(1, num + 1): # choose a value
        # the smallest valid number is 1
        # please notice that num is allowed
        usedint[ind] = val
        # store the results in the array
        # partition the remaining value (num - val)
        helper(num - val, usedint, ind + 1, results)
def partition(num):
    usedint = [0] * num # initialize to 0
    results = list()
    helper(num, usedint, 0, results)
    return results
if __name__ == "__main__":
    num = random.randint(1,10)
    print (f'num = {num}')
    result = partition(num)
    for part in result:
        print (part)
```

```python
# increasing numbers
def helper_increasing(num, usedint, ind, results):
    if (num == 0):
        results.append(usedint[0:ind])
        return
    for val in range(1, num + 1):
        valid = True
        if (ind > 0):
            if (val <= usedint[ind - 1]):
                valid = False
        if (valid):
            usedint[ind] = val
            helper_increasing(num - val, usedint, ind + 1, results)
def partition_increasing(num):
    usedint = [0] * num # initialize to 0
    results = list()
    helper_increasing(num, usedint, 0, results)
    return results
for num in range(2,15):
    print (f'\nnum = {num}')
    result = partition_increasing(num)
    print (f'options = {len(result)}')
    for part in result:
        print (part)
```

```python
# Alternate odd and even numbers
def helper_odd_even(num, usedint, ind, results):
    if (num == 0):
        results.append(usedint[0:ind])
        return
    for val in range(1, num + 1):
        valid = True
        if (ind > 0):
            if ((val % 2) == (usedint[ind - 1] % 2)):
                valid = False
        if (valid):
            usedint[ind] = val
            # store the results in the array
            # partition the remaining value (num - val)
            helper_odd_even(num - val, usedint, ind + 1, results)
def partition_odd_even(num):
    usedint = [0] * num # initialize to 0
    results = list()
    helper_odd_even(num, usedint, 0, results)
    return results
for num in range(2,10):
    print (f'\nnum = {num}')
    result = partition_odd_even(num)
    print (f'options = {len(result)}')
    for part in result:
        print (part)
```
