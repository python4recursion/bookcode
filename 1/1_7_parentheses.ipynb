```python
# Section 1.7.1: print the table
import sys
def g(n, m):
  if (n > m):
    return 0
  if ((n == 0) and (m != 0)):
    return 1
  if ((n == 0) and (m == 0)):
    return 0
  return g(n - 1, m) + g(n, m - 1)
```

```python
import random
def printtable():
  n = random.randint(3,10)
  print(' ',end=' | ')
  for i in range(0, n):
    print(f'{i:5}', end=' ')
  print()
  print('-' * 7 * n)
  for m in range(0, n):
    print(m, end=' | ')
    for i in range(0, n):
      print(f'{g(i, m):5}', end=' ')
    print()
```

```python
# CH02: parentheses.py
import sys
def helper(parentheses, total, left, right, results):
    # parentheses: array to store ( or ) used
    # total: total number of pairs
    # left: how many ( have been used
    # right: how many ) have been used
    ind = left + right
    outstr = '' # empty string
    if (left == total):
        # use up all (
        # remaining are all )
        for i in range(0, ind):
            outstr = outstr + parentheses[i]
        for i in range(right, total):
            outstr = outstr + ')'
        results.append(outstr)
        return
    # next one can always be (
    parentheses[ind] = '('
    helper(parentheses, total, left + 1, right, results)
    # ) can be used only if more ( have been used than )
    if (left > right):
        parentheses[ind] = ')'
        helper(parentheses, total, left, right + 1, results)
def generate(n):
    results = []
    parentheses = ['0'] * (2 * n)
    helper(parentheses, n, 0, 0, results)
    return results
```

```python
import random
def testgenerate():
    n = random.randint(3,9)
    results = generate(n)
    for res in results:
      print(res)
    print(f'n = {n}, {len(results)} results')
testgenerate()
```
