```python
# CH06: buildtree.py
# build a binary tree from postorder and inorder
# not necessary a binary search tree
import random
class BinaryTree:
    def __init__(self, val):
        self.value = val
        self.left = None
        self.right = None
    def helper(self, retlist, order):
        if (order == 'pre'):
            retlist.append(self.value)
        if (self.left != None):
            self.left.helper(retlist, order)
        if (order == 'in'):
            retlist.append(self.value)
        if (self.right != None):
            self.right.helper(retlist, order)
        if (order == 'post'):
            retlist.append(self.value)
    def preorder(self):
        retlist = [] # returned list
        self.helper(retlist, 'pre')
        return retlist
    def postorder(self):
        retlist = [] # returned list
        self.helper(retlist, 'post')
        return retlist
    def inorder(self):
        retlist = [] # returned list
        self.helper(retlist, 'in')
        return retlist
    def insert(self, val):
      if ((random.randint(0, 10000) % 2) == 0): # insert at left
        if (self.left == None):
          self.left = BinaryTree(val)
        else:
          self.left.insert(val)
      else:
        if (self.right == None):
          self.right = BinaryTree(val)
        else:
          self.right.insert(val)
    def print(self, level = 0, prefix = "Root:"):
        # Print right subtree
        if (self.right != None):
            self.right.print(level + 1, "R--")
        # Print root value
        print(" " * 6 * level + prefix, self.value)
        # Print left subtree
        if (self.left != None):
            self.left.print(level + 1, "L--")
```

```python
# construct a binary tree from in-order and post-order
def buildHelper(inorder, postorder, indexes):
    startindex = indexes[0]
    endindex   = indexes[1]
    postindex  = indexes[2]
    if (startindex > endindex):
        return None  # empty list
    val = postorder[postindex]
    tn = BinaryTree(val)
    indexes[2] = indexes[2] - 1
    if (startindex == endindex):
        return tn # no child
    # build the subtrees
    rootindex = inorder.index(val)
    # the root index decrements corretly
    indexes[0] = rootindex + 1  # index after root
    tn.right = buildHelper(inorder, postorder, indexes)
    indexes[0] = startindex     # restore the value
    indexes[1] = rootindex - 1  # index before root
    tn.left = buildHelper(inorder, postorder, indexes)
    return tn
def buildTree(inorder, postorder):
  if (len(inorder) != len(postorder)):
    print('inorder and postorder different legnths')
    return None
  treesize = len(inorder)
  indexes = [0, treesize - 1, treesize - 1]
  # indexes[0]: starting address of the array for a subtree
  # indexes[1]: ending address of the array for a subtree
  # indexes[2]: index for the element used for root
  root = buildHelper(inorder, postorder, indexes)
  return root
```

```python
n = random.randint(40, 100)
k = random.randint(10, int(n / 2))
print(f'n = {n}, k = {k}')
values = random.sample(range(n), k)
print(f'values = {values}, len = {len(values)}')
tree = BinaryTree(values[0])
for i in range(1, len(values)):
    tree.insert(values[i])
preorder1 = tree.preorder()
postorder1 = tree.postorder()
inorder1 = tree.inorder()
print(f'preorder = {preorder1}, len = {len(preorder1)}')
print(f'postorder = {postorder1}, len = {len(postorder1)}')
print(f'inorder = {inorder1}, len = {len(inorder1)}')
tree.print()
```

```python
root = buildTree(inorder1, postorder1)
preorder2 = root.preorder()
postorder2 = root.postorder()
inorder2 = root.inorder()
if (preorder1 != preorder2):
    print('preorder not equal')
if (postorder1 != postorder2):
    print('postorder not equal')
if (inorder1 != inorder2):
    print('inorder not equal')
```
