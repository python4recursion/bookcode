```python
# Section 2.7: binarysearch.py
import random
def helper(arr, low, high, key):
  if (low > high):
      return -1
  mid = int((low + high) / 2)
  if (arr[mid] == key):
      return mid
  if (arr[mid] > key):
      return helper(arr, low, mid - 1, key)
  return helper(arr, mid + 1, high, key)
def binarySearch(arr, key):
  return helper(arr, 0, len(arr) - 1, key)
if __name__ == "__main__":
    length = 15
    arr = []
    for ind in range(0, length):
        arr.append(random.randint(1, 5 * length))
    arr.sort() # binary search needs a sorted array
    print(arr)
    val = random.randint(1, 5 * length)
    print(f'search {val}')
    print(f'index = {binarySearch(arr, val)}')
```
