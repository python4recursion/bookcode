```python
# Section 2.10.2: fibonacci.py
def fib1(n):
    if (n < 1): # invalid
        return 0
    if ((n == 1) or (n == 2)):
        return 1
    return (fib1(n-1) + fib1(n-2))
```

```python
def fib2(n):
    if (n < 1): # invalid
        return 0
    if ((n == 1) or (n == 2)):
        return 1
    a = 1
    b = 1
    for i in range(3, n + 1):
        a, b = b, a + b
    return b
```

```python
from functools import cache
@cache
def fib3(n):
  if (n < 1): # invalid
    return 0
  if ((n == 1) or (n == 2)):
    return n
  else:
    return (fib3(n - 1) + fib3(n - 2))
```

```python
import time
def execution_time(func, r, repeat = 1):
  times = {}
  for n in r:
    start_time = time.time() # Record the start time
    for i in range(repeat):
      val = func(n)
    end_time = time.time() # Record the end time
    times[n] = (end_time - start_time) / repeat
  return times
r = range(11, 41)
time1 = execution_time(fib1, r)
```

```python
time2 = execution_time(fib2, r, 10000)
time3 = execution_time(fib3, r, 10000)
for k in time1:
    print(f'n={k}: t1/t2 = {time1[k] / time2[k]:.6e}, t2/t3 = {time2[k] / time3[k]:.6e}')
```

```python
import matplotlib.pyplot as plt
k_values = list(time1.keys())
plt.figure(figsize=(10, 4))
plt.plot(k_values, list(time1.values()), 'k+', label='fib1')
plt.plot(k_values, list(time2.values()), 'x', label='fib2')
plt.plot(k_values, list(time3.values()), 'm.', label='fib3')
plt.xlabel('Value of n')
plt.ylabel('Time (sec)')
plt.title('Execution Times')
plt.yscale('log') # Add this line to set the y-axis to a logarithmic scale
plt.grid(True)
plt.legend() # Add this line to display the legend
plt.show()
```
