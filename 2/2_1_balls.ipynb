```python
# Section 2.1: balls.py
# f(1) = 2, f(2) = 3, f(n) = f(n-1) + f(n-2)
import sys
def balls(n):
    if (n <= 0):
        print('Invalid argument')
        return -1
    if (n == 1):
        return 2  # f(1) = 2
    if (n == 2):
        return 3  # f(2) = 3
    a = balls(n - 1)
    b = balls(n - 2)
    return (a + b)
if __name__ == "__main__":
    for num in range(2, 11):
      print(f'num = {num}: {balls(num)} options')
```
