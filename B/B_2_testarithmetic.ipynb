```python
# Appendix B.2: arithmetic.py
def myadd(a, b):
    return (a + b)
def mysub(a, b):
    return (a - b)
def mymul(a, b):
    return (a * b) # corrected
```

```python
import unittest
class TestFunc(unittest.TestCase):
    def test_myadd(self):
        self.assertEqual(myadd(3, 5), 8)
    def test_mysub(self):
        self.assertEqual(mysub(13, 5), 8)
if __name__ == "__main__":
    unittest.main(argv=['first-arg-is-ignored'], exit=False)
```
