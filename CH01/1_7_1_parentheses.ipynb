```python
# Section 1.7.1: print the table
import sys
def g(n, m):
  if (n > m):
    return 0
  if ((n == 0) and (m != 0)):
    return 1
  if ((n == 0) and (m == 0)):
    return 0
  return g(n - 1, m) + g(n, m - 1)
```

```python
import random
def printtable():
  n = random.randint(3,10)
  print(' ',end=' | ')
  for i in range(0, n):
    print(f'{i:5}', end=' ')
  print()
  print('-' * 7 * n)
  for m in range(0, n):
    print(m, end=' | ')
    for i in range(0, n):
      print(f'{g(i, m):5}', end=' ')
    print()
printtable()
```
