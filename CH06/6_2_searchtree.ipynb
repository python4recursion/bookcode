```python
# Section 6.2: searchtree.py
# binary search tree
class BinaryTree:
    def __init__(self, val):
        self.value = val
        self.left = None
        self.right = None
    def helper(self, retlist, order):
        if (order == 'pre'):
            retlist.append(self.value)
        if (self.left != None):
            self.left.helper(retlist, order)
        if (order == 'in'):
            retlist.append(self.value)
        if (self.right != None):
            self.right.helper(retlist, order)
        if (order == 'post'):
            retlist.append(self.value)
    def preorder(self):
        retlist = [] # returned list
        self.helper(retlist, 'pre')
        return retlist
    def postorder(self):
        retlist = [] # returned list
        self.helper(retlist, 'post')
        return retlist
    def inorder(self):
        retlist = [] # returned list
        self.helper(retlist, 'in')
        return retlist
    def insert(self, val):
      if (self.value == val): # do not insert the value again
          return
      if (val < self.value):
          if (self.left == None):
              self.left = BinaryTree(val)
              return
          else:
              self.left.insert(val)
              return
      # val > self.value
      if (self.right == None):
            self.right = BinaryTree(val)
            return
      else:
            self.right.insert(val)
    def print(self, level=0, prefix = "Root:"):
        # Print right subtree
        if (self.right != None):
            self.right.print(level + 1, "R--")
        # Print root value
        print(" " * 6 * level + prefix, self.value)
        # Print left subtree
        if (self.left != None):
            self.left.print(level + 1, "L--")
    def search(self, val):
      if (self.value == val):
            return True
      if (self.value > val):
          if (self.left == None):
            return False
          else:
            return self.left.search(val)
      if (self.right == None):
        return False
      return self.right.search(val)
    def delete(self, val):
        if (val < self.value):  # look for value in left subtree
            if (self.left != None):  # left child exists
                self.left = self.left.delete(val)
            return self
        if (val > self.value):
            if (self.right != None):
                self.right = self.right.delete(val)
            return self
        # val equal to self.value
        if ((self.left == None) and (self.right == None)):
            # this node has no child
            return None
        # this node has at least one child
        if (self.left == None):
            return self.right
        if (self.right == None):
            return self.left
        # this node has two children
        # find immediate successor
        imsuc = self.right
        while (imsuc.left != None):
            imsuc = imsuc.left
        # swap the values
        self.value = imsuc.value
        imsuc.value = val
        self.right = self.right.delete(val)
        return self
```

```python
import random
n = random.randint(40, 100)
k = random.randint(10, int(n / 2))
# print(f'n = {n}, k = {k}')
values = random.sample(range(n), k)
values = [8, 4, 11, 1, 9, 7, 10, 5, 12, 15]
k = len(values)
print(f'values = {values}, len = {len(values)}')
tree = BinaryTree(values[0])
for i in range(1, len(values)):
    tree.insert(values[i])
preorder = tree.preorder()
postorder = tree.postorder()
inorder = tree.inorder()
print(f'preorder = {preorder}, len = {len(preorder)}')
print(f'postorder = {postorder}, len = {len(postorder)}')
print(f'inorder = {inorder}, len = {len(inorder)}')
tree.print()
if ((random.randint(0, 10000) % 2 ) == 0):
  v = random.choice(values)
else:
  v = random.randint(0, n * 10)
print(f'search for {v}: {tree.search(v)}')
```
