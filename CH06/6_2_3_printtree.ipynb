```python
# Section 6.2.3: Print a binary tree
class TreeNode:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None
def insert_into_tree(root, value):
    if root is None:
        return TreeNode(value)
    if value < root.value:
        root.left = insert_into_tree(root.left, value)
    else:
        root.right = insert_into_tree(root.right, value)
    return root
def build_binary_tree(values):
    root = None
    for value in values:
        root = insert_into_tree(root, value)
    return root
def print_tree(root, level= 0, prefix="Root:"):
    if root is not None:
        # Print left subtree
        if root.left is not None:
            print_tree(root.left, level + 1, "L--")
        # Print root value
        print("    " * level + prefix, root.value)
        # Print right subtree
        if root.right is not None:
            print_tree(root.right, level + 1, "R--")
```

```python
def inorder_traversal(root):
    if root:
        inorder_traversal(root.left)
        print(root.value, end=" ")
        inorder_traversal(root.right)
```
