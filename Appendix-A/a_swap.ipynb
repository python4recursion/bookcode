```python
 def swap1(a, b):
    print(f'swap1: before a,b = b, a: {a, b}')
    a, b = b, a
    print(f'swap1: after  a, b = b, a: {a, b}')
```

```python
def testswap1():
    x = 2
    y = 19
    print(f'testswap1: before x,y = y, x: {x, y}')
    x, y = y, x
    print(f'testswap1: after x,y = y,x: {x, y}')
    swap1(x, y)
    print(f'testswap1: after calling swap1: {x, y}')
```

```python
class Pair:
    def __init__(self, a, b):
        self.x = a
        self.y = b
    def print(self, arg=''):
        print(f'{arg} Pair.print: {self.x, self.y}')
def swap2(p):
    p.print('swap2: before p.x, p.y = p.y, p.x')
    p.x, p.y = p.y, p.x
    p.print('swap2: after  p.x, p.y = p.y, p.x')
```

```python
def testswap2():
    p = Pair(2, -9)
    p.print('testswap2: before swap2(p)')
    swap2(p)
    p.print('testswap2: after  swap2(p)')
```

```python
testswap1()
print('---')
testswap2()
```
